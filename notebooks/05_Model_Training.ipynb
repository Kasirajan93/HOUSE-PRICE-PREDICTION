# Model Training and Evaluation

## Objective

The objective of this notebook is to train multiple regression models on the preprocessed Ames Housing dataset, evaluate their performance using regression metrics, compare the results, and select the best-performing model for predicting house prices.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso

from sklearn.tree import DecisionTreeRegressor

from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import joblib
```

```python
df = pd.read_csv("../data/processed_housing_data.csv")

df.head()
```

```python
print(df.shape)
```

```python
X = df.drop("SalePrice", axis=1)

y = df["SalePrice"]

print("Features:", X.shape)
print("Target:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print(X_train.shape)
print(X_test.shape)
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

```python
print(X_train.shape)
print(X_test.shape)
```

# Linear Regression

## Objective

Train a Linear Regression model using the training dataset and evaluate its performance on the testing dataset. This model serves as the baseline for comparing the performance of other regression algorithms.

```python
linear_model = LinearRegression()

linear_model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
```

```python
y_pred_linear = linear_model.predict(X_test)

print("First 10 Predictions:")
print(y_pred_linear[:10])
```

```python
linear_mae = mean_absolute_error(y_test, y_pred_linear)

linear_rmse = np.sqrt(mean_squared_error(y_test, y_pred_linear))

linear_r2 = r2_score(y_test, y_pred_linear)

print("Linear Regression Performance")
print("-" * 40)
print(f"Mean Absolute Error (MAE): {linear_mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {linear_rmse:.2f}")
print(f"R² Score: {linear_r2:.4f}")
```

# Linear Regression Summary

The Linear Regression model was successfully trained and evaluated.

### Evaluation Metrics

- **Mean Absolute Error (MAE):** Measures the average prediction error.
- **Root Mean Squared Error (RMSE):** Gives more weight to larger errors.
- **R² Score:** Indicates how well the model explains the variance in house prices.

This model serves as the baseline for comparison with Ridge Regression, Lasso Regression, Decision Tree, and Random Forest.

# Ridge Regression

## Objective

Train a Ridge Regression model to reduce overfitting by applying L2 regularization. The performance will be compared with the baseline Linear Regression model.

```python
ridge_model = Ridge(alpha=1.0)

ridge_model.fit(X_train, y_train)

print("Ridge Regression model trained successfully!")
```

```python
y_pred_ridge = ridge_model.predict(X_test)

print("First 10 Predictions:")
print(y_pred_ridge[:10])
```

```python
ridge_mae = mean_absolute_error(y_test, y_pred_ridge)

ridge_rmse = np.sqrt(mean_squared_error(y_test, y_pred_ridge))

ridge_r2 = r2_score(y_test, y_pred_ridge)

print("Ridge Regression Performance")
print("-" * 40)
print(f"Mean Absolute Error (MAE): {ridge_mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {ridge_rmse:.2f}")
print(f"R² Score: {ridge_r2:.4f}")
```

# Lasso Regression

## Objective

Train a Lasso Regression model using L1 regularization. Lasso can shrink less important feature coefficients to zero, which helps with feature selection and reduces model complexity.

```python
lasso_model = Lasso(alpha=1.0, random_state=42)

lasso_model.fit(X_train, y_train)

print("Lasso Regression model trained successfully!")
```

```python
y_pred_lasso = lasso_model.predict(X_test)

print("First 10 Predictions:")
print(y_pred_lasso[:10])
```

```python
lasso_mae = mean_absolute_error(y_test, y_pred_lasso)

lasso_rmse = np.sqrt(mean_squared_error(y_test, y_pred_lasso))

lasso_r2 = r2_score(y_test, y_pred_lasso)

print("Lasso Regression Performance")
print("-" * 40)
print(f"Mean Absolute Error (MAE): {lasso_mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {lasso_rmse:.2f}")
print(f"R² Score: {lasso_r2:.4f}")
```

# Decision Tree Regression

## Objective

Train a Decision Tree Regressor to capture non-linear relationships between house features and sale price. Compare its performance with Linear, Ridge, and Lasso Regression models.

```python
decision_tree_model = DecisionTreeRegressor(
    random_state=42
)

decision_tree_model.fit(X_train, y_train)

print("Decision Tree Regression model trained successfully!")
```

```python
y_pred_tree = decision_tree_model.predict(X_test)

print("First 10 Predictions:")
print(y_pred_tree[:10])
```

```python
tree_mae = mean_absolute_error(y_test, y_pred_tree)

tree_rmse = np.sqrt(mean_squared_error(y_test, y_pred_tree))

tree_r2 = r2_score(y_test, y_pred_tree)

print("Decision Tree Regression Performance")
print("-" * 40)
print(f"Mean Absolute Error (MAE): {tree_mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {tree_rmse:.2f}")
print(f"R² Score: {tree_r2:.4f}")
```

# Random Forest Regression

## Objective

Train a Random Forest Regressor to improve prediction accuracy by combining multiple Decision Trees. Random Forest generally provides better generalization and is one of the strongest baseline models for structured tabular datasets.

```python
random_forest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

random_forest_model.fit(X_train, y_train)

print("Random Forest Regression model trained successfully!")
```

```python
y_pred_rf = random_forest_model.predict(X_test)

print("First 10 Predictions:")
print(y_pred_rf[:10])
```

```python
rf_mae = mean_absolute_error(y_test, y_pred_rf)

rf_rmse = np.sqrt(mean_squared_error(y_test, y_pred_rf))

rf_r2 = r2_score(y_test, y_pred_rf)

print("Random Forest Regression Performance")
print("-" * 40)
print(f"Mean Absolute Error (MAE): {rf_mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {rf_rmse:.2f}")
print(f"R² Score: {rf_r2:.4f}")
```

# Model Comparison

The performance of all regression models is compared using Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and R² Score.

The model with the highest R² score and the lowest prediction errors is selected as the final model.

```python
comparison_df = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Ridge Regression",
        "Lasso Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "MAE": [
        linear_mae,
        ridge_mae,
        lasso_mae,
        tree_mae,
        rf_mae
    ],
    "RMSE": [
        linear_rmse,
        ridge_rmse,
        lasso_rmse,
        tree_rmse,
        rf_rmse
    ],
    "R2 Score": [
        linear_r2,
        ridge_r2,
        lasso_r2,
        tree_r2,
        rf_r2
    ]
})

comparison_df = comparison_df.sort_values(
    by="R2 Score",
    ascending=False
)

comparison_df
```
