# Model Interpretation

## Objective

Interpret the trained Random Forest model by identifying the most important features that influence house prices.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib
```

```python
model = joblib.load("../models/best_model.pkl")

print("Best model loaded successfully!")
```

```python
df = pd.read_csv("../data/processed_housing_data.csv")

print(df.shape)
```

```python
X = df.drop("SalePrice", axis=1)
```

```python
importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

importance.head(20)
```

```python
plt.figure(figsize=(10,8))

plt.barh(
    importance["Feature"][:20][::-1],
    importance["Importance"][:20][::-1]
)

plt.title("Top 20 Important Features")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()
```

# Interpretation

The Random Forest model identified the most influential variables affecting house prices.

Typical important features include:

- Overall Quality
- Ground Living Area
- Garage Capacity
- Total Basement Area
- Year Built

These features contribute the most to the prediction of house prices.
