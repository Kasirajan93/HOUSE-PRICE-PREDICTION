# House Price Prediction System

## Objective

Load the trained Random Forest model and StandardScaler to predict the sale price of houses.

This notebook verifies that the saved model can successfully make predictions before deployment.

```python
import pandas as pd
import numpy as np
import joblib
```

```python
model = joblib.load("../models/best_model.pkl")

scaler = joblib.load("../models/scaler.pkl")

print("Model loaded successfully!")

print("Scaler loaded successfully!")
```

```python
df = pd.read_csv("../data/processed_housing_data.csv")

print(df.shape)

df.head()
```

```python
X = df.drop("SalePrice", axis=1)

y = df["SalePrice"]

print("Features Shape:", X.shape)

print("Target Shape:", y.shape)
```

```python
sample_house = X.iloc[[0]]

actual_price = y.iloc[0]

print("Actual House Price:")

print(actual_price)
```

```python
sample_scaled = scaler.transform(sample_house)
```

```python
predicted_price = model.predict(sample_scaled)

print("Predicted House Price:")

print(predicted_price[0])
```

```python
comparison = pd.DataFrame({
    "Actual Price": [actual_price],
    "Predicted Price": [predicted_price[0]]
})

comparison
```

```python
error = abs(actual_price - predicted_price[0])

percentage_error = (error / actual_price) * 100

print(f"Absolute Error : {error:.2f}")

print(f"Percentage Error : {percentage_error:.2f}%")
```

# Prediction Summary

The saved Random Forest model successfully predicted the sale price of a sample house.

## Workflow

- Loaded the trained model
- Loaded the StandardScaler
- Loaded the processed dataset
- Selected a sample house
- Scaled the input
- Predicted the house price
- Compared the prediction with the actual sale price
